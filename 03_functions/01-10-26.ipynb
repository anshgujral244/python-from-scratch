```python
# A function is a reusable block of code that performs a task.
def greet():
    print("hello world ")

# Call the function by writing its name followed by parentheses.
greet()
greet()
```

```python
# name is a parameter; "Ansh" is the argument passed to the function.
def greet(name):
    print(f"hello {name}!")

greet("Ansh")
```

```python
# return sends a value back to the caller; print only displays a value.
def add(first_number, second_number):
    return first_number + second_number

result = add(5, 7)
print(result)
```

```python
# If a function has no return statement, Python returns None automatically.
def no_return():
    value = 5

print(no_return())
```

```python
#default arguments
def power(base, exponent=2):
    return base**exponent

print(power(3))  # Uses default exponent of 2
print(power(3, 3))  # Uses exponent of 3

#
```

```python
#⚠️ Never use a mutable default like def f(items=[]). The list is created once and shared across calls. Use None instead:
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

```python
#positional and keyword arguments
def describe(name,role,city):
    return f"{name} is a {role} from {city}."

print(describe("Ansh","developer","New York"))  # Positional arguments
print(describe(name="Ansh", role="developer", city="New York"))  #
```

```python
# *args and **kwargs
def total(*args):
    return (args)
    return sum(args)

print(total(1, 2, 3, 4, 5))
```

```python
def profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

profile(name="Ansh", skill="SQL", level="intermediate")
```

```python
#multiple return values
def min_max(numbers):
    return min(numbers), max(numbers)

min_val, max_val = min_max([1, 2, 3, 4, 5])
print(f"Minimum: {min_val}, Maximum: {max_val}")
```

```python
#scope:local and global variables
x = 10   # global

def change():
    x = 99          # a NEW local variable
    print("inside:", x)

change()
print("outside:", x)
```

```python
#docstrings and type hints
def are(length: float, width: float) -> float:
    """Calculate the area of a rectangle."""
    return length * width

print(are(5.0,7.0))
help(are)
```

```python
#lambda functions
square = lambda x: x ** 2
print(square(5))

nums=[3,1,4,1,5]
print(sorted(nums, key=lambda n: -n))
```

```python
# functions are objects 
def apply(func, values):
    return [func(v) for v in values]

print(apply(str.upper, ["sql", "python"]))
print(list(map(lambda x: x * 2, [1, 2, 3])))
print(list(filter(lambda x: x % 2 == 0, range(10))))
```

```python
#recursion 
def fact(n):
    if n == 0:
        return 1
    return n * fact(n - 1)
print(fact(5)) 
```
