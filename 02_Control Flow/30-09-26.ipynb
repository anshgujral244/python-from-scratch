```python
marks = 75

if marks >= 60:
    print("Pass")
else:
    print("Fail")
```

```python
#we use elif when we have multiple conditions to check
marks = 85

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
#indentation is important in python. It is used to define the blocks of code. The code inside the if, elif, and else statements must be indented.
x = 10

if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```
