```python
count =1

while count <=5:
    print(count)
    count+=1
```

```python
Why this works
count starts at 1
loop runs while count <= 5
after each loop, count increases by 1
when count becomes 6, the condition becomes false and the loop stops
```

```python
num = 5

while num > 0:
    print(num)
    num -= 1
```

```python
# infinite loop danger
x = 1

while x < 5:
    print(x)
    # x is never changed here
```

```python
#using a while loop to ask for a password
password = "python"

user_input = ""

while user_input != password:
    user_input = input("Enter password: ")

print("Access granted")
```

```python
Difference between for and while
for loop: used when you know how many times or want to iterate through items.
while loop: used when the number of iterations depends on a condition.
```
