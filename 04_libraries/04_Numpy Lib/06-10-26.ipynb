```python
#working with csv data files
```

```python
climate_data=np.genfromtxt('climate.txt',delimiter=',',skip_header=1)
```

```python
climate_data
```

```python
climate_data.shape
```

```python
yeilds = climate_data @ weights
```

```python
yeilds
```

```python
#joining two arrays using concatenate function. It is used to join two or more arrays along a specified axis. The concatenate function takes a tuple of arrays as input and returns a new array that is the concatenation of the input arrays.
climate_data= np.concatenate((climate_data,yeilds.reshape(10000,1)),axis=1)
```

```python
climate_data
```

```python
np.savetxt('climate_results.txt',climate_data,fmt='%.2f',header='temperature,rainfall,humidity,yeilds_apples',comments='')
```
