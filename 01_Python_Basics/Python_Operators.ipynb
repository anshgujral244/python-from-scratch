## 1. Arithmetic Operators

Arithmetic operators calculate with numbers. `/` performs division, `//` performs floor division, `%` gives a remainder, and `**` raises a number to a power.

```python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 3)
```

**Try it:** Change the values below. What does the remainder tell you about whether a number is even?

```python
number = 14
print(number % 2)  # 0 means even; 1 means odd
```

## 2. Comparison Operators

Comparisons return `True` or `False`. Use `==` to compare values and `=` to assign a value. Chained comparisons are useful for ranges.

```python
score = 82
print(score == 82)
print(score != 100)
print(score >= 50)
print(18 <= 21 < 65)
```

## 3. Logical Operators

`and` requires both conditions to be true, `or` requires at least one, and `not` reverses a truth value.

```python
age = 22
has_ticket = True
print(age >= 18 and has_ticket)
print(age < 18 or has_ticket)
print(not has_ticket)
```

Python uses short-circuit evaluation: the right side of `and` or `or` may not run if the result is already known. For example, the list access below only happens if the list is not empty.

```python
items = ["apple", "pear"]
print(len(items) > 0 and items[0] == "apple")
```

## 4. Assignment Operators

Augmented assignment updates a variable using its current value: `x += 3` is equivalent to `x = x + 3`. Other forms include `-=`, `*=`, `/=`, `//=`, `%=`, and `**=`.

```python
points = 10
points += 5
points *= 2
print(points)
```

## 5. Membership Operators

`in` checks whether a value is present; `not in` checks that it is absent. For a dictionary, membership checks keys.

```python
skills = ["python", "sql", "excel"]
print("sql" in skills)
print("r" not in skills)
person = {"name": "Asha", "city": "Delhi"}
print("name" in person)
print("Asha" in person.values())
```

## 6. Identity Operators

`==` compares values. `is` checks whether two names refer to the same object. Use `is None` to check for `None`; for ordinary values, prefer `==`.

```python
first = [1, 2]
second = [1, 2]
third = first
print(first == second)  # Equal contents
print(first is second)  # Different list objects
print(first is third)   # Same list object
result = None
print(result is None)
```

## 7. Bitwise Operators

Bitwise operators act on the binary digits of integers. They are less common in everyday beginner programs. `&` is AND, `|` is OR, `^` is XOR, `~` inverts bits, and `<<` / `>>` shift bits. Do not confuse these with logical `and` and `or`.

```python
print(5 & 3)   # 1
print(5 | 3)   # 7
print(5 ^ 3)   # 6
print(5 << 1)  # 10
print(5 >> 1)  # 2
print(~5)      # -6
```

## 8. Operator Precedence

Multiplication happens before addition. Parentheses make the order explicit. When an expression is hard to read, use parentheses. Exponentiation groups from right to left.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3 ** 2)
print((2 ** 3) ** 2)
```

A useful simplified order is: parentheses, exponentiation, unary operators, multiplication/division/remainder, addition/subtraction, comparisons and membership, `not`, `and`, then `or`. Python's complete precedence rules include more cases; parentheses are your friend.

## 9. Common Mistakes

- Use `==` to compare, not `=`.
- `/` returns a float; use `//` for floor division.
- Use `==` for ordinary value comparisons; reserve `is` for identity checks such as `value is None`.
- Use `and` and `or` for logical conditions; `&` and `|` are bitwise operators.
- Division by zero raises `ZeroDivisionError`.
- Add parentheses when they make the intended calculation clearer.
